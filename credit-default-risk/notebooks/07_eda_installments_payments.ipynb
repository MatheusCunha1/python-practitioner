```python
import duckdb
from utils import create_view   

con = duckdb.connect()
create_view(con)
```

```python
con.sql("""        
    DESC VW_INSTALLMENTS_PAYMENTS
""").show()
```

```python
con.sql("""
    SELECT COUNT(*) FROM VW_INSTALLMENTS_PAYMENTS             
""").show()
```

```python
con.sql("""
    SUMMARIZE VW_INSTALLMENTS_PAYMENTS
""").show()
```

```python
consulta = con.sql("""
    SELECT 
        'SELECT ' || COLUMN_NAME ||', COUNT(*) FROM VW_INSTALLMENTS_PAYMENTS GROUP BY ' || COLUMN_NAME ||' ORDER BY 2 DESC LIMIT 5' AS CONSULTA
    FROM duckdb_columns()
    WHERE UPPER(TABLE_NAME) = 'VW_INSTALLMENTS_PAYMENTS'
""")

select = consulta.fetchall()

for i, in select: 
    con.sql(i).show()
```
