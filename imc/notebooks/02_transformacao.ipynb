# Transformação: conversão de unidades
Como o dataset importado pelo Kaggle estava em unidades diferentes, como polegadas e libras, é necessário fazer transformações para lidarmos com esses dados.

```python
import pandas as pd
import numpy as np 

df = pd.read_csv(r"..\data\data_frame_kaggle.csv")

df = df.rename(columns={
    'Gender': 'Sexo',
    'Height': 'Altura',
    'Weight': 'Peso'
})

df['Sexo'] = df['Sexo'].replace({
    'Male': 'M',
    'Female': 'F'
})

df['Altura'] =  df['Altura'] * 0.0254
df['Peso'] =  df['Peso'] * 0.4536 

df['Altura'] = (np.trunc(df['Altura'] * 100) / 100)
df['Peso'] = (np.trunc(df['Peso'] * 100) / 100)
```

```python
print(df)
```

```python
df.to_csv(r"..\data\data_frame_transformado.csv", index=False)
```
